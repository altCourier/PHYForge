```python
import matplotlib.pyplot as plt # alias is the right word to call this

import numpy as np
import pandas as pd

# Displaying plots inline!
%matplotlib inline

"""
As per the source code states:

- One of the style names in `.style.available` (a builtin style or
              a style installed in the user library path).

"""
plt.style.use('default')

plt.style.available
```

Key concepts:

* Figure is the top level container
* Axes the actual plot area where the data is drawn
* Axis is the number of line obj (x, y, z etc.)

Figure -> canvas, Axes -> individual paintings on that canvas.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Pyplot approach:

plt.plot(x, y)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')

plt.title('Simple Line Plot')

plt.plot(y,y)

plt.show() # No need to type this as we already did the inline above!
           # This is also the part where we can show it.

plt.plot(x,x)
```

```python
# Instead of using the global plt object, it is also possible to use
# subplots

# This makes it easier to organize them

fig, ax = plt.subplots() # Creates new obj

ax.plot(x, y)

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Line Plot')

# fig.show() # What is the difference between fig.show AND plt.show in here?
plt.show()
```

```python
x = np.linspace(0, 10, 100) # 100 points from 0 to 10
y = np.sin(x)

plt.figure(figsize = (8, 5)) # W = 8, H = 5

plt.plot(x, y)

plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.title('Sine Wave')

plt.grid(True) # Add grid for easier reading

plt.show()
```

```python
# Another example of multiple lines on the plot

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure( figsize = (10, 6) )
plt.plot(x, y1, label = 'sin(x)')
plt.plot(x, y2, label = 'cos(x)')
plt.plot(x, y3, label = 'sin(x) * cos(x)')

plt.xlabel('X values')
plt.ylabel('Y values')

plt.title(' Multiple Functions ')

plt.legend()

plt.grid(True, alpha = 0.3) # Alpha = transparency!

plt.show()
```

```python
# Customization and Other Charts

x = np.linspace(0, 10, 20)

plt.figure(figsize = (12, 6))

"""

Basically colors are:

b = blue, r = red, c = cyan, g = green, m = magenta, y = yellow, k = black!

markers: 

o = circle, s = square, ^ = triangle, * = star, + = plus, x = x

line styles:

- = solid, -- = dashed, -. = dash-dot, : = dotted

"""


plt.plot(x, x, 'b-', label = 'Solid line', linewidth = 2)
plt.plot(x, x + 1, 'r--', label = 'Dashed line', linewidth = 2)
plt.plot(x, x + 2, 'c-.', label = 'dash-dot line', linewidth = 2)
plt.plot(x, x + 3, 'g:', label = 'Dotted line', linewidth = 2)
plt.plot(x, x + 4, 'yo-', label = 'Circles', linewidth = 2)
plt.plot(x, x + 5, 'ms-', label = 'Squares', linewidth = 2)
plt.plot(x, x + 6, 'k^-', label = 'Triangles', linewidth = 2)

plt.title('')

plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.legend()

plt.grid(True, alpha = 0.4)
```

Instead of passing these arguments into parameters ambiguously, it is entirely possible to also give them explicitly.

```python
x = np.linspace(0, 10, 100)

plt.figure(figsize = (10, 6))

plt.plot(x, np.sin(x), color = 'steelblue', linewidth = 2, label = 'steelblue')
plt.plot(x, np.cos(x), color = "#9a2727", linewidth = 2, label = '#9a2727 (hex)')
plt.plot(x, np.sin(x) * 0.5, color = 'darkgreen', linewidth = 2, linestyle = '--' ,label = 'darkgreen')

plt.xlabel('X Values', fontsize = 12)
plt.ylabel('Y Values', fontsize = 12)

plt.title('Custom Color and Styles', fontsize = 14, fontweight = 'bold')

plt.legend(fontsize = 10)

plt.grid(True, alpha = 0.6)

plt.show()
```

```python
# Scatter Plots

np.random.seed()

x = np.random.randn(100)
y = 2 * x * np.random.rand(100) * 0.5

plt.figure( figsize = (8,6) )
plt.scatter(x, y, alpha = 0.6)

plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.title('Scatter Plot')
plt.grid(True, alpha = 0.6)

plt.show()
```

```python
# Scatter plot w different colors and sizes

np.random.seed(42)

x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.randn(100)
sizes = 1000 * np.random.randn(100)

plt.figure( figsize = (10, 6) )

plt.scatter(x, y, c = colors, s = sizes, alpha = 0.6, cmap = 'viridis')
plt.colorbar(label = 'Color instensity') # Add colorbar

plt.xlabel('X values')
plt.ylabel('Y values')

plt.title('Scatter plots w color and size variations')

plt.grid(True, alpha = 0.3)

plt.show()
```

```python
# Bar Charts

categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 101, 12]

plt.figure( figsize = (12, 10))

plt.bar(categories, values, color = 'darkred', edgecolor = 'brown', linewidth = 1.2)
plt.xlabel('Categories')
plt.ylabel('Values')

plt.title('Simple Bar Chart')

plt.grid(True, alpha = 0.3, axis = 'y')

plt.show()

```

```python
# Horizontal Bar Chart

plt.figure(figsize = (12, 6))
plt.barh(categories, values, color = 'coral', edgecolor = 'black', linewidth = 1.2)
 # The only change is the h in the barh function!
plt.xlabel('Values')
plt.ylabel('Categories')

plt.title('Horizontal Bar Chart')

plt.grid(True, alpha = 0.3, axis = 'x')

plt.show()
```

```python
# Grouped Bar Chart

categories = ['Q1', 'Q2', 'Q3', 'Q4']

product_a = [20, 35, 30, 35]
product_b = [25, 32, 34, 20]
product_c = [30, 25, 30, 25]

x = np.arange(len(categories))
width = 0.25


plt.figure(figsize = (10, 6))

plt.bar(x - width, product_a, width, label = 'Product A', color = '#FF6B6B')
plt.bar(x, product_b, width, label = 'Product B', color = '#43CDC4')
plt.bar(x + width, product_c, width, label = 'Product C', color = '#95E1D3')

plt.xlabel('Quarter')
plt.ylabel('Sales')

plt.title('Quarterly Sales Comparison')

plt.xticks(x, categories)

plt.legend()

plt.grid(True, alpha = 0.3, axis = 'y')

plt.show()
```
