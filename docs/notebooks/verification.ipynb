```python
import torch
from sionna.phy.mapping import Mapper, Demapper 

# 1. Setup a simple QPSK-2 bit mapper and demapper
mapper = Mapper("qam", 2)
demapper = Demapper("app", "qam", 2)

# 2. Create a known bit sequence: [0, 0, 0, 1, 1, 0, 1, 1]
bits = torch.tensor([[0, 0, 0, 1, 1, 0, 1, 1]], dtype = torch.float32)

# 3. Map to symbols, then immediately demap
# N0 = 0.001 to avoid division by zero in the demapper
symbols = mapper(bits)
llrs = demapper(symbols, 0.001)

# 4. Compare input bits to output LLR signs
print("Input Bits:", bits.tolist())
print("Output LLRs:", torch.round(llrs, decimals=2).tolist())
```

```python
from sionna.phy.mapping import Constellation

modulations = [
    ("pam", 1),  # BPSK?
    ("qam", 2),  # QPSK
    ("qam", 4),  # 16-QAM
    ("qam", 6),  # 64-QAM
    ("qam", 8),  # 256-QAM
    ("qam", 10)  # 1024-QAM
]

for mod_type, bits_per_symbol in modulations:
    try:
        constellation = Constellation(mod_type, bits_per_symbol)
        num_points = constellation.points.shape[0]
        print(f"Type: {mod_type}, Bits/Sym: {bits_per_symbol} -> Constellation Size: {num_points}")
    except Exception as e:
        print(f"Failed to create {mod_type} with {bits_per_symbol} bits/symbol: {e}")
```
