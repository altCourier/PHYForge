```python
import numpy as np
from sionna.phy.mapping import qam

def dump_labeled_constellation(num_bits_per_symbol: int, normalize: bool = True):
    points = qam(num_bits_per_symbol, normalize=normalize)
    print(f"{'index':>6} | {'bit label':^{num_bits_per_symbol}} | {'I':>8} | {'Q':>8}")
    print("-" * 40)
    for i, point in enumerate(points):
        label = np.binary_repr(i, num_bits_per_symbol)
        print(f"{i:>6} | {label:^{num_bits_per_symbol}} | {point.real:>8.4f} | {point.imag:>8.4f}")

# 16-QAM
dump_labeled_constellation(4)

print()

# 64-QAM
dump_labeled_constellation(6)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sionna.phy.mapping import qam

def plot_constellation(num_bits_per_symbol: int, normalize: bool = True):
    points = qam(num_bits_per_symbol, normalize=normalize)
    
    plt.figure(figsize=(6, 6))
    plt.scatter(points.real, points.imag, marker='o', s=50, color='b')
    
    # Aesthetics
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)
    plt.title(f'{2**num_bits_per_symbol}-QAM Constellation')
    plt.xlabel('In-Phase (I)')
    plt.ylabel('Quadrature (Q)')
    plt.axis('equal') # Important: keeps the aspect ratio correct
    plt.show()

# 16-QAM
plot_constellation(4)

# 64-QAM
plot_constellation(6)
```
